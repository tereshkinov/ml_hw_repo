```python
class Solution:
    def searchInsert(self, nums: list[int], target: int) -> int:
        pos = len(nums) // 2
        if len(nums) == 0: return 0
        if target == nums[pos]: return pos
        if target < nums[pos]: return self.searchInsert(nums[:pos], target)
        if target > nums[pos]: return pos + self.searchInsert(nums[pos + 1:], target)
```

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def reverseList(head: ListNode | None) -> ListNode | None:
    pos = 0
    lst = head
    ptr = lst.next
    while ptr != None:
        ptr_next = ptr.next
        ptr.next = lst
        lst = ptr
        ptr = ptr_next
    return lst

```
